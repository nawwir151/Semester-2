```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []

    def add_child(self, child_node):
        self.children.append(child_node)

    def __str__(self, level=0):
        # Membuat indentasi visual menggunakan spasi
        ret = "  " * level + str(self.data) + "\n"
        for child in self.children:
            ret += child.__str__(level + 1)
        return ret

# --- Contoh Penggunaan ---
print("Membuat Struktur Tree Sederhana:")

# 1. Membuat node root
root = TreeNode("Global")

# 2. Membuat node wilayah
region_asia = TreeNode("Asia")
region_europe = TreeNode("Europe")

root.add_child(region_asia)
root.add_child(region_europe)

# 3. Menambahkan negara ke wilayah
region_asia.add_child(TreeNode("Indonesia"))
region_asia.add_child(TreeNode("Japan"))
region_europe.add_child(TreeNode("Germany"))
region_europe.add_child(TreeNode("France"))

# 4. Mencetak hasil
print(root)
```

```python
class BinaryNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    def insert_left(self, data):
        if self.left is None:
            self.left = BinaryNode(data)
        else:
            new_node = BinaryNode(data)
            new_node.left = self.left
            self.left = new_node

    def insert_right(self, data):
        if self.right is None:
            self.right = BinaryNode(data)
        else:
            new_node = BinaryNode(data)
            new_node.right = self.right
            self.right = new_node

    def delete(self, data):
        # Mencari dan menghapus anak kiri jika datanya cocok
        if self.left and self.left.data == data:
            self.left = None
            return True
        # Mencari dan menghapus anak kanan jika datanya cocok
        if self.right and self.right.data == data:
            self.right = None
            return True
        
        # Rekursi ke bawah jika tidak ditemukan di level ini
        res = False
        if self.left:
            res = self.left.delete(data)
        if not res and self.right:
            res = self.right.delete(data)
        return res
        
    def __str__(self, level=0, prefix="Root: "):
        ret = " " * level + prefix + str(self.data) + "\n"
        if self.left:
            ret += self.left.__str__(level + 1, "L--- ")
        if self.right:
            ret += self.right.__str__(level + 1, "R--- ")
        return ret
```

```python
def inorder_traversal(node):
    """ Traversasi Inorder (Left -> Root -> Right). """
    if node:
        inorder_traversal(node.left)
        print(node.data, end=" ")
        inorder_traversal(node.right)

def preorder_traversal(node):
    """ Traversasi Preorder (Root -> Left -> Right). """
    if node:
        print(node.data, end=" ")
        preorder_traversal(node.left)
        preorder_traversal(node.right)

def postorder_traversal(node):
    """ Traversasi Postorder (Left -> Right -> Root). """
    if node:
        postorder_traversal(node.left)
        postorder_traversal(node.right)
        print(node.data, end=" ")
```

```python
class ExpNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def evaluate(node):
    # Jika leaf node (angka), kembalikan nilainya
    if node.left is None and node.right is None:
        return int(node.value)
    
    # Evaluasi sub-tree kiri dan kanan
    left_val = evaluate(node.left)
    right_val = evaluate(node.right)
    
    # Terapkan operator
    if node.value == '+':
        return left_val + right_val
    if node.value == '-':
        return left_val - right_val
    if node.value == '*':
        return left_val * right_val
    if node.value == '/':
        return left_val / right_val
```
