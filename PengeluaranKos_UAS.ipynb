```python
# ===========================================================
# BLOK 1 : STRUKTUR DATA NODE & DOUBLE LINKED LIST
# ===========================================================
# Materi Struktur Data:
# - Node
# - Doubly Linked List (Double Linked List)
#
# Fungsi:
# Menyimpan setiap transaksi sebagai node yang saling terhubung
# menggunakan pointer next dan prev.
#
# Struktur:
# HEAD <-> NODE 1 <-> NODE 2 <-> NODE 3 <-> TAIL
#
# Digunakan sebagai dasar implementasi Stack untuk menyimpan
# riwayat transaksi pengeluaran.
# ===========================================================

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class LogTransaksiStackLL:
    def __init__(self):
        self.head = None
        self.tail = None
        self._size = 0


    # ===========================================================
    # BLOK 2 : IMPLEMENTASI STACK (LIFO)
    # ===========================================================
    # Materi Struktur Data:
    # - Stack (Last In First Out)
    #
    # Fungsi:
    # push()        -> Menambah transaksi terbaru
    # pop_terbaru() -> Menghapus transaksi terbaru
    # lihat_log()   -> Menampilkan seluruh transaksi
    #
    # Implementasi Stack menggunakan Double Linked List
    # dengan tail sebagai TOP OF STACK.
    #
    # Konsep:
    # Data terakhir masuk akan menjadi data pertama keluar.
    # ===========================================================

    def push(self, hari, kategori, nominal):
        log_data = {"hari": hari, "kategori": kategori, "nominal": nominal}
        new_node = Node(log_data)

        if self.tail is None:
            self.head = self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

        self._size += 1

    def pop_terbaru(self):
        if self.is_empty():
            return None

        temp = self.tail

        if self.head == self.tail:
            self.head = self.tail = None
        else:
            self.tail = self.tail.prev
            self.tail.next = None

        self._size -= 1
        return temp.data

    def is_empty(self):
        return self.tail is None

    def lihat_log(self):
        if self.is_empty():
            print("-> Belum ada riwayat pengeluaran bulan ini.")
            return

        print("Riwayat Pengeluaran Harian (Kronologis - Struktur Linked List):")

        current = self.head
        indeks = 1

        while current is not None:
            log = current.data
            print(f"  [{indeks}] Hari ke-{log['hari']} | Kategori: {log['kategori']} | Sebesar: Rp{log['nominal']:,}")
            current = current.next
            indeks += 1


# ===========================================================
# BLOK 3 : PROGRAM UTAMA (MAIN PROGRAM)
# ===========================================================
# Fungsi:
# Mengatur seluruh alur aplikasi Expense Tracker.
#
# Di dalam blok ini digunakan beberapa struktur data:
#
# 1. Dictionary (Hash Table)
#    -> Menyimpan informasi saldo dan pengeluaran.
#
# 2. List
#    -> Menyimpan daftar kategori pengeluaran.
#
# 3. Array 2 Dimensi (Matriks)
#    -> Menyimpan rekap pengeluaran harian selama 30 hari.
#
# 4. Stack berbasis Linked List
#    -> Menyimpan log transaksi dan mendukung fitur Undo.
# ===========================================================

def main():

    print("|.....................................................|")
    print("         PENGATURAN AWAL KEUANGAN ANAK KOS             ")
    print("|.....................................................|")

    while True:
        try:
            uang_bulanan = int(input("MASUKKAN TOTAL UANG SAKU SEBULAN (Rp): "))
            if uang_bulanan < 0:
                print("UANG SAKU TIDAK BOLEH MINUS!")
                continue
            break
        except ValueError:
            print("[INPUT SALAH] Harap masukkan angka saja tanpa titik/huruf!")


    # ===========================================================
    # BLOK 4 : DICTIONARY (HASH TABLE)
    # ===========================================================
    # Materi Struktur Data:
    # - Dictionary
    # - Hash Table
    #
    # Fungsi:
    # Menyimpan data keuangan utama program.
    #
    # Key:
    # saldo_awal
    # total_pengeluaran
    # sisa_saldo
    #
    # Akses data menggunakan key sehingga pencarian data
    # berlangsung cepat.
    # ===========================================================

    dompet_kos = {
        "saldo_awal": uang_bulanan,
        "total_pengeluaran": 0,
        "sisa_saldo": uang_bulanan
    }


    # ===========================================================
    # BLOK 5 : LIST
    # ===========================================================
    # Materi Struktur Data:
    # - List
    #
    # Fungsi:
    # Menyimpan daftar kategori pengeluaran.
    #
    # Indeks:
    # 0 -> Makan & Minum
    # 1 -> Transportasi
    # 2 -> Keperluan Lainnya
    # ===========================================================

    daftar_kategori = [
        "Makan & Minum",
        "Transportasi",
        "Keperluan Lainnya"
    ]


    # ===========================================================
    # BLOK 6 : ARRAY 2 DIMENSI (MATRIKS)
    # ===========================================================
    # Materi Struktur Data:
    # - Array 2D
    # - Matriks
    #
    # Ukuran:
    # 30 Baris x 3 Kolom
    #
    # Baris:
    # Hari ke-1 sampai Hari ke-30
    #
    # Kolom:
    # [0] Makan & Minum
    # [1] Transportasi
    # [2] Keperluan Lainnya
    #
    # Fungsi:
    # Menyimpan total pengeluaran berdasarkan hari
    # dan kategori.
    # ===========================================================

    tabel_harian_2d = [[0 for _ in range(3)] for _ in range(30)]


    # ===========================================================
    # BLOK 7 : PEMBUATAN OBJEK STACK
    # ===========================================================
    # Membuat objek Stack berbasis Double Linked List
    # untuk menyimpan riwayat transaksi.
    # ===========================================================

    log_jajan = LogTransaksiStackLL()


    while True:

        print("\n|.......................................|")
        print("     PERFECT EXPENSE TRACKER SYSTEM      ")
        print("|.......................................|")

        print("1. Lihat Dashboard Ringkasan & Matriks Tabel 30 Hari")
        print("2. Input Pengeluaran Harian Baru (Tambah Transaksi)")
        print("3. Hapus Catatan Pengeluaran TERBARU (Koreksi Salah Input)")
        print("4. Laporan Akhir Bulan (Evaluasi & Sisa Uang)")
        print("5. Keluar Aplikasi")

        pilihan = input("Pilih menu (1-5): ")


        # ===========================================================
        # MENU 1 : DASHBOARD KEUANGAN
        # ===========================================================
        # Menampilkan:
        # - Data Dictionary
        # - Data Array 2D
        # - Data Linked List
        # ===========================================================

        if pilihan == "1":
            pass


        # ===========================================================
        # MENU 2 : INPUT TRANSAKSI
        # ===========================================================
        # Proses:
        #
        # 1. Input data dari pengguna
        # 2. Push ke Stack Linked List
        # 3. Simpan ke Array 2D
        # 4. Update Dictionary
        #
        # Struktur Data yang terlibat:
        # - Stack
        # - Linked List
        # - Array 2D
        # - Dictionary
        # ===========================================================

        elif pilihan == "2":
            pass


        # ===========================================================
        # MENU 3 : HAPUS TRANSAKSI TERBARU
        # ===========================================================
        # Struktur Data:
        # - Stack (LIFO)
        #
        # Method:
        # pop_terbaru()
        #
        # Fungsi:
        # Menghapus transaksi terakhir yang dimasukkan
        # kemudian memperbarui Array 2D dan Dictionary.
        # ===========================================================

        elif pilihan == "3":
            pass


        # ===========================================================
        # MENU 4 : LAPORAN AKHIR BULAN
        # ===========================================================
        # Menampilkan:
        # - Total Modal Awal
        # - Total Pengeluaran
        # - Sisa Saldo
        # - Persentase Sisa Uang
        #
        # Digunakan untuk evaluasi pengelolaan keuangan.
        # ===========================================================

        elif pilihan == "4":
            pass


        # ===========================================================
        # MENU 5 : KELUAR PROGRAM
        # ===========================================================
        # Mengakhiri eksekusi aplikasi.
        # ===========================================================

        elif pilihan == "5":
            break


# ===========================================================
# BLOK 8 : ENTRY POINT PROGRAM
# ===========================================================
# Menjalankan fungsi utama ketika file dieksekusi.
# ===========================================================

if __name__ == "__main__":
    main()
```
